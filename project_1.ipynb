```python
import csv

zodiac_s = []

with open("profiles_revised.csv", newline="", encoding="utf-8") as f:
    reader_s = csv.reader_s(f)
    header_s = next(reader_s)
    for row_s in reader_s:
        zodiac_s.append(dict(zip(header_s, row_s)))
```

`csv.reader()` is a function in the `csv` module that returns "a reader object that will process lines from the given csvfile" (Python Software Foundation, n.d.-a).  I describe it as a row-by-row iterator returning a list of strings.  `next()` is a built-in function that takes an iterator and returns its next item (Python Software Foundation, n.d.-b) in this case, the header row.  `dict()` is a built-in function that takes key-value pairs (or nothing) and builds a dictionary (Python Software Foundation, n.d.-c).  

`zip()` is a built-in function that pairs up items from two or more iterables by position and returns a zip object, an iterator that gives out those pairs as tuples (Python Software Foundation, n.d.-d).  I describe the zip object as a pair-by-pair iterator that gives out a tuple each time it's looped over.  I also had to learn the `encoding="utf-8"` parameter, which I describe as a guard against garbling.
