## INSTRUCTION 6: real work under Standard Library (SD)

### inputing under SD...done (below)

We have open() from Homework 3 but I want to learn new tools from the library for concise coding.  I consulted Claude on four designs.  The most parsimonious design uses csv.reader(), next(), dict(), and zip() which we have not learned yet.  I present a hand-written versions of the two most ambitious code blocks for the standard library work in Appendix C.  I wrote down by hand twice for both meaning and structure.

I understand csv.reader() as a row-by-row iterator that each time returns a list of strings, next() as a built-in function that takes an iterator and returns its next item in this case, the header row, and dict() build a dictionary for key-value pairs where the keys are the column names and the values are one row's entries.  I know understand it pulls more info than I want or need.  I am, however, committed to the design as studied it. 

I describe the zip object as a pair-by-pair iterator that gives out a tuple each time it's looped over.  I also had to learn the encoding="utf-8" parameter, which I describe as a guard against garbling. 

```python
import csv

zodiac_s = []

with open("profiles_revised.csv", newline="", encoding="utf-8") as f:
    reader_s = csv.reader(f)
    header_s = next(reader_s)
    for row_s in reader_s:
        zodiac_s.append(dict(zip(header_s, row_s)))
```

## INSTRUCTION 6 continued

### statistical work (median)...done (below)

This design is very consistent what we have worked on in class so far.  sorted() produces new data.  len() gives the length.  We then find the middle position through floor division and then account for an even number length. 

```python
sorted_s = sorted(ages_s)
n = len(sorted_s)
middle = n // 2
if n % 2 == 1:
    median_s = sorted_s[middle]
else:
    median_s = (sorted_s[middle - 1] + sorted_s[middle]) / 2
```

## INSTRUCTION 6 continued

### statistical work (mode)...done (below)

counts_s is a lookup table and a dictionary.  We will give it key-value pairs.  I think of it as an mx2 matrix but I understand that is math metaphor and imprecise heresy otherwise.  The iterator, on the first iteration, is empty so we add 1 to get the blood moving.  max() iterates across the keys and compares their counts.  As of the below, INSTRUCTION 5 is done. 

```python
counts_s = {}
for a in ages_s:
    counts_s[a] = counts_s.get(a, 0) + 1
mode_s = max(counts_s, key=counts_s.get)
print(mode_s)
```

## INSTRUCTION 7 continued

### visualization...done (below)

```python
top = max(counts_s.values())
print("Age Distribution of OkCupid Users, (SF, 2012)")
print("Age | Users (each * =", top // 36, "users)")
for age in sorted(counts_s):
    print(age, "*" * (counts_s[age] * 36 // top))
```
