// inputing under Standard Library (SD)

We have open() from Homework 3 but I want see other tools (rather than heavy-looping) from the library for concise coding.  I consulted Claude on four designs.  The most parsimonious design uses csv.reader(), next(), dict(), and zip() which we have not learned yet.  I used the four-step AI-usage guidance and also present a hand-written version of the design in Appendix C.

```python
import csv

zodiac_s = []

with open("profiles_revised.csv", newline="", encoding="utf-8") as f:
    reader_s = csv.reader(f)
    header_s = next(reader_s)
    for row_s in reader_s:
        zodiac_s.append(dict(zip(header_s, row_s)))
```

Pandas 

pandas: read_csv already turns empty cells and common placeholders like "NA", "null" and "NaN" into NaN. So isna() catches most of it. The one thing it misses is cells that contain only spaces, which stay as strings.


csv module: it never produces nulls (None). Every value is a string, so check for empty strings ''. Also check for placeholder text like "NA" and for cells that are only spaces (.strip() handles those).




pandas: It reads types for you, turns blanks and “NA” into NaN, and cleans with one-liners like isna(), dropna() and fillna().
The hard way: Everything is a string. You check for blanks and placeholders yourself, convert types with int(), and clean row by row in a loop.

`csv.reader()` is a function in the `csv` module that returns "a reader object that will process lines from the given csvfile" (Python Software Foundation, n.d.-a).  I describe it as a row-by-row iterator returning a list of strings.  `next()` is a built-in function that takes an iterator and returns its next item (Python Software Foundation, n.d.-b) in this case, the header row.  `dict()` is a built-in function that takes key-value pairs (or nothing) and builds a dictionary (Python Software Foundation, n.d.-c).  

`zip()` is a built-in function that pairs up items from two or more iterables by position and returns a zip object, an iterator that gives out those pairs as tuples (Python Software Foundation, n.d.-d).  I describe the zip object as a pair-by-pair iterator that gives out a tuple each time it's looped over.  I also had to learn the `encoding="utf-8"` parameter, which I describe as a guard against garbling.
